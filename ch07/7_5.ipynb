```python
# 고객 군집합

import pandas as pd
import datetime
import math
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel(io = "Online_Retail.xlsx")
df.head()
```

```python
df = df[df["Quantity"]>0] #음수 드랍
df = df[df["UnitPrice"]>0]
df = df[df["CustomerID"].notnull()] # 결측치 제거
df.shape
```

```python
df["Country"].value_counts()[:5]
```

```python
df = df[df["Country"] == "United Kingdom"]
df.shape
```

```python
df["sale_amount"] = df["Quantity"]*df["UnitPrice"]
```

```python
df.shape
```

```python
df.info()
```

```python
df["CustomerID"] = df [ "CustomerID"].astype(int)
```

```python
df.groupby(["InvoiceNo", "StockCode",])["InvoiceNo"].count().mean()
```

```python
aggregations = {
    "InvoiceDate" : "max",
    "InvoiceNo" : "count",
    "sale_amount" : "sum"
}

cust_df = df.groupby("CustomerID").agg(aggregations)

cust_df = cust_df.rename(columns={"InvoiceDate": "Recency","InvoiceNo": "Frequency","sale_amount": "Monetary"})

cust_df = cust_df.reset_index()
cust_df.head()
```

```python
import datetime as dt
cust_df["Recency"] = dt.datetime(2011, 12 ,10) - cust_df["Recency"]
cust_df["Recency"] = cust_df["Recency"].apply(lambda x : x.days +1)
print(cust_df.shape)
cust_df.head()
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(figsize=(12, 4), nrows=1, ncols=3)

ax1.set_title("Recency")
ax1.hist(cust_df["Recency"])

ax2.set_title("Frequency")
ax2.hist(cust_df["Frequency"])

ax3.set_title("Monetary")
ax3.hist(cust_df["Monetary"])

plt.tight_layout()
```

```python
cust_df[["Recency","Frequency","Monetary"]].describe()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X_features = cust_df[["Recency","Frequency","Monetary"]].values
X_features_scaled = StandardScaler().fit_transform(X_features)

Kmeans = KMeans(n_clusters= 3 ,random_state= 0)
labels = Kmeans.fit_predict(X_features_scaled)
cust_df["cluster_label"] = labels

print(np.round(silhouette_score(X_features_scaled, labels), 3))
```

```python
from sklearn.metrics import silhouette_samples, silhouette_score

def visualize_silhouette(cluster_list, X_features):
    """여러 군집 수에 대한 실루엣 그래프를 나란히 그립니다."""
    import matplotlib.cm as cm

    n_cols = len(cluster_list)
    fig, axes = plt.subplots(
        1, n_cols, figsize=(4 * n_cols, 5), squeeze=False
    )
    axes = axes.ravel()

    for ind, n_clusters in enumerate(cluster_list):
        if n_clusters < 2 or n_clusters >= len(X_features):
            raise ValueError("군집 수는 2 이상, 데이터 샘플 수 미만이어야 합니다.")

        clusterer = KMeans(
            n_clusters=n_clusters,
            max_iter=500,
            n_init=10,
            random_state=0
        )
        cluster_labels = clusterer.fit_predict(X_features)
        sil_avg = silhouette_score(X_features, cluster_labels)
        sil_values = silhouette_samples(X_features, cluster_labels)

        ax = axes[ind]
        ax.set_title(
            f"군집 수: {n_clusters}\n평균 실루엣 점수: {sil_avg:.3f}"
        )
        ax.set_xlabel("실루엣 계수")
        ax.set_ylabel("군집")
        ax.set_xlim([-0.1, 1.0])
        ax.set_xticks(np.arange(-0.1, 1.1, 0.2))
        ax.set_yticks([])

        y_lower = 10
        for cluster_id in range(n_clusters):
            cluster_values = sil_values[cluster_labels == cluster_id]
            cluster_values.sort()
            cluster_size = cluster_values.shape[0]
            y_upper = y_lower + cluster_size
            color = cm.nipy_spectral(float(cluster_id) / n_clusters)

            ax.fill_betweenx(
                np.arange(y_lower, y_upper),
                0,
                cluster_values,
                facecolor=color,
                edgecolor=color,
                alpha=0.7
            )
            ax.text(-0.05, y_lower + 0.5 * cluster_size, str(cluster_id))
            y_lower = y_upper + 10

        ax.axvline(x=sil_avg, color="red", linestyle="--", linewidth=1)

    fig.tight_layout()
    plt.show()
```

```python
def visualize_kmeans_plot_multi(cluster_lists, X_features):
    from sklearn.decomposition import PCA

    n_cols = len(cluster_lists)
    fig, axs = plt.subplots(
        nrows=1, ncols=n_cols, figsize=(4 * n_cols, 4), squeeze=False
    )
    axs = axs.ravel()

    pca = PCA(n_components=2)
    pca_transformed = pca.fit_transform(X_features)
    dataframe = pd.DataFrame(pca_transformed, columns=["PCA1", "PCA2"])
    markers = ["o", "s", "^", "x", "*"]

    for ind, n_cluster in enumerate(cluster_lists):
        cluster = KMeans(
            n_clusters=n_cluster, max_iter=500, n_init=10, random_state=0
        )
        cluster_labels = cluster.fit_predict(pca_transformed)
        dataframe["cluster"] = cluster_labels

        for label in np.unique(cluster_labels):
            label_df = dataframe[dataframe["cluster"] == label]
            axs[ind].scatter(
                label_df["PCA1"],
                label_df["PCA2"],
                s=70,
                edgecolors="k",
                marker=markers[label % len(markers)],
                label=f"Cluster {label}"
            )

        axs[ind].set_title(f"Number of clusters: {n_cluster}")
        axs[ind].set_xlabel("PCA1")
        axs[ind].set_ylabel("PCA2")
        axs[ind].legend()

    fig.tight_layout()
    plt.show()
```

```python
visualize_kmeans_plot_multi([2, 3, 4, 5], X_features_scaled)
```

```python
cust_df["Recency_log"] = np.log1p(cust_df["Recency"])
cust_df["Frequency_log"] = np.log1p(cust_df["Frequency"])
cust_df["Monetary_log"] = np.log1p(cust_df["Monetary"])

X_features_log = cust_df[["Recency_log", "Frequency_log", "Monetary_log"]].values
X_features_log_scaled = StandardScaler().fit_transform(X_features_log)

kmeans = KMeans(n_clusters=3, random_state=0, n_init=10)
labels = kmeans.fit_predict(X_features_log_scaled)
cust_df["Cluster_label_log"] = labels

print(np.round(silhouette_score(X_features_log_scaled, labels), 3))
```
