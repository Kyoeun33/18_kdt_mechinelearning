```python
import sklearn
from sklearn.datasets import load_iris 
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

```

```python
import pandas as pd

iris = load_iris()

iris_data = iris.data # 피처 : 꽃받힘의 너비 길이 , 꽃잎의 너비 길이
iris_label = iris.target # 아이리스 품종 : 정답

print("iris target: " , iris_label)
print("iris target 명 : ", iris.target_names)
```

```python
iris.data
```

```python
iris_df = pd . DataFrame(data=iris_data,columns=iris.feature_names)
iris_df["label"] = iris.target

iris_df.head()
```

```python
# 학습셋과 테스트 셋으로 나누기

X_train, X_test, Y_train, Y_test = train_test_split(iris_data, iris_label, test_size=0.2,train_size=0.8, random_state=1) 
```

```python
dt_clf = DecisionTreeClassifier(random_state=1)
```

```python
dt_clf.fit(X_train,Y_train)
```

```python
pred = dt_clf.predict(X_test)
```

```python
pred
```

```python
from sklearn.metrics import accuracy_score
print("예측 정확도 : ", accuracy_score(Y_test, pred))
```
