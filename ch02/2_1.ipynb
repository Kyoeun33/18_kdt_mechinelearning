```python
import sklearn
from sklearn.datasets import load_iris 
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

```

```python
# 학습셋과 테스트 셋으로 나누기

X_train, X_test, Y_train, Y_test = train_test_split(iris_data, iris_label, test_size=0.2,train_size=0.8, random_state=1) 
```

```python
dt_clf = DecisionTreeClassifier(random_state=1)
```

```python
dt_clf.fit(X_train,Y_train)
```

```python
pred = dt_clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("예측 정확도 : ", accuracy_score(Y_test, pred))
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np

n_iter = 0
cv_accuracy = []

features = iris.data
label = iris.target

dt_clf = DecisionTreeClassifier(random_state=1)

for train_index, test_index in skf.split(features, label):
    X_train, X_test = features[train_index], features[test_index]
    Y_train, Y_test = label[train_index], label[test_index]

    # 모델 학습
    dt_clf.fit(X_train, Y_train)

    # 검증 데이터 예측
    pred = dt_clf.predict(X_test)

    # 정확도 계산
    accuracy = accuracy_score(Y_test, pred)
    cv_accuracy.append(accuracy)

    n_iter += 1
    print(f"{n_iter}번째 Fold 정확도: {accuracy:.4f}")

print(f"\n평균 교차 검증 정확도: {np.mean(cv_accuracy):.4f}")
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(dt_clf, features, label, scoring = "accuracy",cv = skf)

print("교차 검증별 정확도 : ", np.round(scores , 4))
print("평균 검증 정확도 : ", np.round(np.mean(scores) , 4))
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.2,
    random_state=121,
    stratify=iris.target
)

parameters = {
    "max_depth": [1, 2, 3],
    "min_samples_split": [2, 3]
}
```

```python

from sklearn.model_selection import GridSearchCV

dtree = DecisionTreeClassifier(random_state=156)

grid_dtree = GridSearchCV(
    estimator=dtree,
    param_grid=parameters,
    cv=3,
    scoring="accuracy",
    refit=True
)

grid_dtree.fit(X_train, Y_train)

scores_df = pd.DataFrame(grid_dtree.cv_results_)

scores_df[
    [
        "params",
        "mean_test_score",
        "rank_test_score",
        "split0_test_score",
        "split1_test_score",
        "split2_test_score"
    ]
]
```
