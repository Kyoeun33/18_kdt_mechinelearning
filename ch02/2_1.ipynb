train_test_split() : 머신러닝에서 데이터를 학습용 데이터와 테스트용 데이터로 나누는 함수

> 훈련 데이터(Train): 문제와 정답을 보면서 공부하는 데이터
> 테스트 데이터(Test): 공부에 사용하지 않고, 실력을 평가하는 데이터

>> test_size = 0.2 ==> 전체의 20%를 테스트용으로 사용

>> random_state = 아무 숫자 ==>	실행할 때마다 같은 방식으로 분할하도록 설정

>>>X_train :훈련용 문제 (모델이 학습할 꽃의 특징 데이터)
>>>X_test : 테스트용 문제 (학습에 사용하지 않고 평가할 꽃의 특징 데이터)
>>>y_train : 훈련용 정답 (훈련 데이터에 대응하는 품종 번호)
>>>y_test : 테스트용 정답 (모델의 예측 결과가 맞는지 비교할 실제 정답)

((대문자 X는 여러 특징을 가진 2차원 데이터, 소문자 y는 보통 1차원 정답 데이터)) ##근데 이미 난 대문자 Y로 해버려서,,,

KFold : 데이터를 여러 묶음으로 나누어 모델을 반복적으로 학습하고 평가할 수 있게 해주는 도구

>> KFold(n_splits = 5)  n_splits = 의 수만큼 묶음을 만듬
>> cv_accuracy = []	각 반복에서 측정한 정확도를 저장할 빈 리스트

구분	        train_test_split()     	  KFold--> 결과적으로 모든 데이터가 테스트데이터로 사용됨
분할 방식     	한 번 나눔	             여러 번 나눔
테스트 데이터	 보통 한 번 사용     	각 묶음이 한 번씩 사용
목적	       기본적인 학습·평가    	평가 결과의 안정성 확인
결과	       한 번의 평가 결과        	여러 평가 결과

```python
from sklearn.model_selection import KFold
import numpy as np

Kfold = KFold(n_splits = 5)
cv_accuracy = []
```

```python
n_iter = 0  # 반복 횟수를 세기 위한 변수
features = iris.data #붓꽃의 4가지 측정값
label = iris.target #붓꽃의 실제 품종 번호
cv_accuracy = [] #각 반복의 정확도를 저장할 빈 리스트

# Kfold.split(features)--> 인덱스 배열 2개를 반환
# train_index: 학습에 사용할 데이터의 행 번호
# test_index: 검증에 사용할 데이터의 행 번호


for train_index, test_index in Kfold.split(features):
    X_train, X_test = features[train_index],features[test_index]
    Y_train, Y_test = label[train_index],label[test_index]

    dt_clf.fit(X_train,Y_train)
    pred = dt_clf.predict(X_test)
    n_iter += 1

    accuracy = np.round(accuracy_score(Y_test,pred),4)

    train_size = X_train.shape[0]
    test_size =X_test.shape[0]
    print("교차검증 정확도 : ",accuracy)
    print("학습 데이터 크기 : ",train_size)
    print("검증데이터크기 : ",test_size)
    cv_accuracy.append(accuracy)
    
print("평균 검증 정확도",np.mean(cv_accuracy))
```

```python
#####################

from sklearn.model_selection import StratifiedKFold

# 레이블 비율을 유지하는 3-Fold 교차 검증
skf = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=1
)

n_iter = 0

for train_index, test_index in skf.split(
    iris_df.drop(columns=["label"]),
    iris_df["label"]
):
    n_iter += 1

    label_train = iris_df["label"].iloc[train_index]
    label_test = iris_df["label"].iloc[test_index]

    print(f"\n[{n_iter}번째 Fold]")
    print("학습 레이블 데이터 분포:")
    print(label_train.value_counts().sort_index())

    print("검증 레이블 데이터 분포:")
    print(label_test.value_counts().sort_index())
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np

n_iter = 0
cv_accuracy = []

features = iris.data
label = iris.target

dt_clf = DecisionTreeClassifier(random_state=1)

for train_index, test_index in skf.split(features, label):
    X_train, X_test = features[train_index], features[test_index]
    Y_train, Y_test = label[train_index], label[test_index]

    # 모델 학습
    dt_clf.fit(X_train, Y_train)

    # 검증 데이터 예측
    pred = dt_clf.predict(X_test)

    # 정확도 계산
    accuracy = accuracy_score(Y_test, pred)
    cv_accuracy.append(accuracy)

    n_iter += 1
    print(f"{n_iter}번째 Fold 정확도: {accuracy:.4f}")

print(f"\n평균 교차 검증 정확도: {np.mean(cv_accuracy):.4f}")
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(dt_clf, features, label, scoring = "accuracy",cv = skf)

print("교차 검증별 정확도 : ", np.round(scores , 4))
print("평균 검증 정확도 : ", np.round(np.mean(scores) , 4))
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.2,
    random_state=121,
    stratify=iris.target
)

parameters = {
    "max_depth": [1, 2, 3], # 의사결정나무의 최대 깊이 후보
    "min_samples_split": [2, 3]# 노드를 분할하기 위한 최소 샘플 수 후보
}
```

```python

from sklearn.model_selection import GridSearchCV

# 의사결정나무 분류 모델 생성
dtree = DecisionTreeClassifier(random_state=156)

grid_dtree = GridSearchCV(
    estimator=dtree,
    param_grid=parameters,
    cv=3,
    scoring="accuracy",
    refit=True
)

grid_dtree.fit(X_train, Y_train)

# 각 하이퍼파라미터 조합의 교차 검증 결과를 데이터프레임으로 변환
scores_df = pd.DataFrame(grid_dtree.cv_results_)

scores_df[
    [
        "params",
        "mean_test_score",
        "rank_test_score",
        "split0_test_score",
        "split1_test_score",
        "split2_test_score"
    ]
]
```

```python
# 조합을 교차 검증으로 평가하는 GridSearchCV 불러오기
from sklearn.model_selection import GridSearchCV

dtree = DecisionTreeClassifier(random_state=156)
# 조합별 모델 성능을 비교하도록 설정
grid_dtree = GridSearchCV(
    estimator=dtree,
    param_grid=parameters,
    cv=3,
    scoring="accuracy",
    refit=True
)

grid_dtree.fit(X_train, Y_train)
# 탐색 및 교차 검증 실행
scores_df = pd.DataFrame(grid_dtree.cv_results_)

scores_df[
    [
        "params",
        "mean_test_score",
        "rank_test_score",
        "split0_test_score",
        "split1_test_score",
        "split2_test_score"
    ]
]
```
